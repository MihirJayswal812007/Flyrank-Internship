## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is a Binary Classification task. The model will look at a page's metadata, format, and current performance, and classify whether it is fundamentally failing at user engagement (Needs Redesign = 1) or performing as expected (Healthy = 0).

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The true target is "user frustration with the page layout," which we cannot measure directly. My proxy target will be pages that rank high (e.g., page 1 or top 3) but suffer from abnormally low Click-Through Rates (CTR) compared to their peers.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The success metric will be Precision@K (e.g., Precision@50). If the model hands the design team a queue of 50 comparison articles to redesign, we want a high percentage of them to actually be pages bleeding traffic due to format issues, so we don't waste engineering hours.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

# Load the starter dataset
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Unit of analysis: 1 row = 1 webpage / URL
print(f"Unit of analysis: 1 row = 1 anonymized page. Total pages: {len(df)}")

# Sketching the target column (Proxy: "needs_redesign")
# Let's flag "comparison" pages on page 1 that have a terrible CTR (< 5%)
df["needs_redesign"] = np.where(
    (df["content_type"].str.contains("comparison", case=False, na=False)) &
    (df["position_tier"] == "page_1") &
    (df["ctr"] < 0.05),
    1, 0
)

# Show the unit of analysis and our new sketched target column (Removed url_id)
display(df[["content_type", "position_tier", "ctr", "needs_redesign"]].head(10))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (like "flag all pages with CTR < 5%") is too rigid. It ignores the nuance of different content formats. An ML model can weigh content_type, content_age_days, word_count, and position_tier simultaneously to recognize complex patterns of structural decay that a simple hardcoded threshold would miss.
