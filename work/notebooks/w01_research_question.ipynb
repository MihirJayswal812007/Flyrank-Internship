## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Make the filter case-insensitive and flexible
comparison_df = df[df["content_type"].str.contains("comparison", case=False, na=False)]

# Get some quick numbers for our framing
total_pages = len(df)
comparison_pages = len(comparison_df)
avg_ctr_comparison = comparison_df["ctr"].mean()
avg_ctr_overall = df["ctr"].mean()

print(f"Total pages in dataset: {total_pages}")
print(f"Number of 'comparison' format pages: {comparison_pages}")
print(f"Average CTR for comparison pages: {avg_ctr_comparison:.3f} (vs {avg_ctr_overall:.3f} overall)")
```
